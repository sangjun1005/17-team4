```python
import pandas as pd

df = pd.read_excel(r"C:\Users\김양일\Desktop\Team_Project\data\raw\02_의료기관.xlsx", header=1)

print(df.head())
print(df.columns)
```

```python
print(df.info())
```

```python
print(df.isnull().sum())
```

```python
df2 = df.rename(
    columns={
        "읍면동별(2)" :"지역",
        "합계": "의료기관수"
    }
)

print(df2.head())
```

```python
df2 = df2[["지역", "의료기관수", "보건소", "보건지소", "보건진료소"]].copy()

print(df2.head(10))
print(df2.shape)
```

```python
print(df2.info())
```

```python
print(df2.isnull().sum())
```

```python
df2 = df2.drop(index=0).reset_index(drop=True)
```

```python
df2["의료기관수"] = df2["의료기관수"].replace("-", 0)
df2["의료기관수"] = df2["의료기관수"].astype(int)
df2["보건소"] = df2["보건소"].replace("-", 0)
df2["보건소"] = df2["보건소"].astype(int)
df2["보건지소"] = df2["보건지소"].replace("-", 0)
df2["보건지소"] = df2["보건지소"].astype(int)
df2["보건진료소"] = df2["보건진료소"].replace("-", 0)
df2["보건진료소"] = df2["보건진료소"].astype(int)

print(df2.info())
print(df2.head())
```

```python
df_med = df2[["지역", "의료기관수", "보건지소", "보건진료소"]].copy()

print(df_med.head())
print(df_med.shape)
```

```python
df = pd.read_csv(r"C:\Users\김양일\Desktop\Team_Project\data\processed\01_인구_v3.csv")

print(df.head())
```

```python
df_region = df[["지역", "65세이상", "고령인구비율"]].copy()

print(df_region.head())
print(df_region.shape)
```

```python
df_merge = pd.merge(df_region, df_med, on="지역", how= "inner")

print(df_merge.head())
```

```python
df_merge["고령인구1000명당의료기관수"] = (df_merge["의료기관수"] / df_merge["65세이상"] * 1000).round(2)

print(df_merge.head())
```

```python
df5 = df_merge.sort_values(["의료기관수","보건지소","보건진료소"]).head(10)
print(df5[["지역","의료기관수","보건지소","보건진료소"]])
```

```python
df_sort = df_merge.sort_values("고령인구1000명당의료기관수", ascending=True)

df_sort.to_csv(r"C:\Users\김양일\Desktop\Team_Project\data\processed\02_의료기관1_v2.csv",
            index=False,
            encoding="utf-8-sig")

print(df_sort.head(10))
```

```python
result = df_merge[df_merge["65세이상"] >= 3000].sort_values("고령인구1000명당의료기관수")

print(result[["지역", "65세이상", "의료기관수", "고령인구1000명당의료기관수"]].head(10))
```
