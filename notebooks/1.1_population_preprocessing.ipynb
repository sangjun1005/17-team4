```python
file_path = r"C:\Users\김양일\Desktop\팀프로젝트\data\raw\연령별인구현황(2026.6월).xlsx"
sheet_name = "4.연령별(10세 등)"

df = pd.read_excel(file_path, sheet_name=sheet_name, header=None)

print(df.head())
print(df.shape)
```

```python
df = pd.read_excel(file_path, sheet_name=sheet_name, header = 2)

print(df.head(10))
print(df.columns)
```

```python
df = df.rename(columns={
    "연령" : "지역",
    "Unnamed: 1" : "성별"
    })

print(df.columns)
```

```python
df["지역"] = df["지역"].ffill()

print(df.head())
```

```python
df["지역"] = df["지역"].str.replace(" ", "", regex=False)

print(df[["지역", "성별"]].head(12))
```

```python
df_total = df[df["성별"] == "계"].copy()
df_total = df_total.reset_index(drop=True)

print(df_total.head(10))
print(df_total.shape)
```

```python
print(df_total.isnull().sum())
```

```python
print(df_total.head(10))
print(df_total.shape)
```

```python
df_total["고령인구비율"] = round(df_total["65세이상"] / df_total["총계"] *100, 2)

df_total = df_total.sort_values("고령인구비율", ascending=False)
print(df_total.head(10))
```

```python
display(df_total[["지역", "총계", "고령인구비율"]].head(10))
```
