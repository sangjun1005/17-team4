```python
import pandas as pd

df = pd.read_csv(r"C:\Users\김양일\Desktop\Team_Project\data\processed\01_인구_v2.csv")

print(df.head())
```

```python
df["85세이상비율"] = (df["85세이상"] / df["총계"] * 100).round(2)
print(df.head())
```

```python
import matplotlib.pylab as plt
import matplotlib

matplotlib.rc("font", family="Malgun Gothic")
plt.rcParams["axes.unicode_minus"] = False

df_old85_top10 = df.sort_values("85세이상비율", ascending=False).head(10)

plt.figure(figsize=(8, 4))
plt.bar(
    df_old85_top10["지역"],
    df_old85_top10["85세이상비율"],
)
plt.title("지역별 85세 이상 비율 TOP10")
plt.xlabel("지역")
plt.ylabel("85세 이상 비율(%)")
plt.xticks(rotation=70)
plt.grid(axis="y", alpha=0.5)

plt.show()
```
