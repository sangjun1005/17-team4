```python
import pandas as pd
import matplotlib.pyplot as plt


plt.rc('font', family='Malgun Gothic')
plt.rcParams['axes.unicode_minus'] = False


data = {
    "지역": ["동면", "수신면", "성남면", "광덕면", "일봉동", "북면", "풍세면", "봉명동", "성거읍", "원성2동"],
    "의료기관수": [0, 0, 0, 0, 1, 2, 5, 6, 6, 7],
    "보건지소": [1, 1, 1, 1, 0, 1, 1, 0, 1, 0],
    "보건진료소": [1, 1, 2, 2, 0, 1, 2, 0, 0, 0]
}

df = pd.DataFrame(data)

df["공공보건시설"] = df["보건지소"] + df["보건진료소"]

df_plot = df.set_index("지역")[["의료기관수", "공공보건시설"]]
df_plot.columns = ["민간 의료기관", "공공 보건시설"]

ax = df_plot.plot(
    kind="bar", 
    figsize=(12, 8), 
    width=0.7, 
    color=['#4C72B0', '#55A868'] 
)

plt.title("민간 의료기관 및 공공 보건시설 수가 적은 지역 10곳", fontsize=14, pad=15, fontweight='bold')
plt.xlabel("지역", fontsize=12, fontweight='bold')
plt.ylabel("시설 수 (개소)", fontsize=12, fontweight='bold')
plt.xticks(rotation=0)  
plt.legend(title="구분", fontsize=10)
plt.grid(axis='y', linestyle='--', alpha=0.5)

for container in ax.containers:
    ax.bar_label(container, fmt='%d', padding=3, fontsize=9)

plt.ylim(0, df_plot.values.max() + 2)

plt.tight_layout()
plt.savefig("02_민간 의료기관 및 공공 보건시설 수가 적은 지역 10곳.png", dpi=300, bbox_inches="tight")

plt.show()
```
