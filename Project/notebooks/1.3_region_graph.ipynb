```python
import pandas as pd
import matplotlib.pyplot as plt

plt.rc('font', family='Malgun Gothic')
plt.rcParams['axes.unicode_minus'] = False

data = {
    "지역": ["동면", "성남면", "수신면", "광덕면", "북면", "중앙동", "입장면", "병천면", "원성2동", "성환읍"],
    "65세이상": [1020, 1328, 946, 1857, 1726, 1660, 2750, 2056, 2856, 6823],
    "고령인구비율": [54.72, 48.57, 47.73, 47.01, 43.41, 37.77, 37.53, 35.22, 33.63, 31.84]
}

df = pd.DataFrame(data)

fig, ax1 = plt.subplots(figsize=(12, 8))

color1 = '#4C72B0'
bars = ax1.bar(df["지역"], df["65세이상"], color=color1, alpha=0.7, width=0.5, label="65세이상 인구수")
ax1.set_xlabel("지역", fontsize=12, fontweight='bold')
ax1.set_ylabel("65세이상 인구수 (명)", color=color1, fontsize=12, fontweight='bold')
ax1.tick_params(axis='y', labelcolor=color1)
ax1.grid(axis='y', linestyle='--', alpha=0.4)

ax1.bar_label(bars, fmt='%d명', padding=3, fontsize=9)

ax2 = ax1.twinx()
color2 = '#C44E52'
line = ax2.plot(df["지역"], df["고령인구비율"], color=color2, marker='o', linewidth=2, label="고령인구비율(%)")
ax2.set_ylabel("고령인구비율 (%)", color=color2, fontsize=12, fontweight='bold')
ax2.tick_params(axis='y', labelcolor=color2)


for i, txt in enumerate(df["고령인구비율"]):
    ax2.annotate(f"{txt}%", (df["지역"][i], df["고령인구비율"][i]), 
                 textcoords="offset points", xytext=(0, 8), ha='center', 
                 fontsize=9, color=color2, fontweight='bold')

plt.title("지역별 65세이상 인구수 및 고령인구비율 현황", fontsize=14, pad=15, fontweight='bold')
ax2.set_ylim(20, 65)  

plt.savefig("지역별 65세이상 인구수 및 고령인구비율 현황.png", dpi=300, bbox_inches="tight")

plt.tight_layout()
plt.show()
```
