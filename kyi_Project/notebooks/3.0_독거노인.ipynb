```python
import pandas as pd

df = pd.read_excel(r"C:\Users\김양일\Desktop\Team_Project\data\raw\03_독거노인_현황_연령별.xlsx", header= 1)

print(df.head())
```

```python
print(df.info())
```

```python
print(df.isnull().sum())
```

```python
print(df)
```

```python
df = df.drop([0, 1, 2, 20])
```

```python
print(df)
```

```python
df = df.reset_index(drop=True)
```

```python
print(df.head())
print(df.shape)
```

```python
df = df[["구별(2)", "합계"]].copy()
print(df.head())
print(df.shape)
```

```python
df = df.rename(columns={
    "구별(2)": "지역",
    "합계": "독거노인수"
}
)
print(df.head())
print(df.columns)
```

```python
print(df.info())
print(df.isnull().sum())
print(df["독거노인수"].unique())
```

```python
df["독거노인수"] = df["독거노인수"].astype(int)

print(df.info())
```

```python
df["지역"] = df["지역"].replace(" ", "")
print(df.head())
print(df.shape)
```

```python
df_merged = pd.read_csv(r"C:\Users\김양일\Desktop\Team_Project\data\processed\Result_인구+의료.csv")
```

```python
print(df_merged.head())
```

```python
df_merged = df_merged.rename(columns={"행정구역":"지역"})
print(df_merged.head())
```

```python
miss1 = sorted(set(df_merged["지역"]) - set(df["지역"]))

print(miss1)
```

```python
miss2 = sorted(set(df["지역"]) - set(df_merged["지역"]))

print(miss2)
```

```python
df_merged = df_merged.merge(
    df,
    on="지역",
    how="left"
)

print(df_merged.head())
print(df_merged.isnull().sum())
```

```python
df_merged["고령인구대비 독거노인비율"] = (df_merged["독거노인수"] / df_merged["65세이상"] * 100).round(2)

print(df_merged.head())
```

```python
print(df.isnull().sum())
```

```python
df_merged.to_csv(r"C:\Users\김양일\Desktop\Team_Project\data\processed\Result_인구+의료+독거노인.csv",
                 index=False,
                 encoding="utf-8-sig")
```

```python
top10 = df_merged[["지역", "독거노인수"]].sort_values("독거노인수", ascending=False).head(10)
print(top10)
```

```python
import matplotlib.pyplot as plt
import matplotlib

matplotlib.rc("font", family="Malgun Gothic")
plt.rcParams["axes.unicode_minus"] = False

plt.figure(figsize=(10, 5))
plt.bar(top10["지역"], top10["독거노인수"], color = "#55A868")
plt.xticks(rotation=45)
plt.title("독거노인 수 TOP10")
plt.tight_layout()
plt.show()
```
